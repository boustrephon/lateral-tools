# Global Earthquake Model (GEM)

To extract seismic hazard data for a specific location using the **Global Earthquake Model (GEM)**, follow these steps:

### **1. Access GEM Tools & Platforms**
GEM provides several tools and datasets for seismic hazard assessment:
- **[GEM Hazard Model](https://www.globalquakemodel.org/gem-hazard)** – Offers global and regional seismic hazard maps.
- **[OpenQuake Platform](https://platform.openquake.org/)** – Provides access to hazard and risk calculations.
- **[OpenQuake Engine](https://github.com/gem/oq-engine)** – An open-source tool for running custom hazard and risk analyses.

### **2. Using the OpenQuake Platform**
#### **Option A: Interactive Hazard Maps**
1. Go to **[GEM’s OpenQuake Platform](https://platform.openquake.org/)**.
2. Navigate to **"Hazard"** section.
3. Select the relevant **hazard model** (e.g., Global Hazard Mosaic, regional models).
4. Use the map interface to zoom into your location of interest.
5. Click on the site to get:
   - **Peak Ground Acceleration (PGA)**
   - **Spectral Acceleration (SA) values** for different return periods (e.g., 475-year, 2475-year).
6. Export data in CSV or GeoJSON format.

#### **Option B: Run a Custom Hazard Analysis**
If you need site-specific hazard curves or uniform hazard spectra:
1. **Download OpenQuake Engine** (if running locally) or use the **OpenQuake Web Tools**.
2. Prepare an **input file** (INI format) specifying:
   - Location (latitude/longitude or region)
   - Hazard model (e.g., `global` or a regional model)
   - Intensity measure types (PGA, SA at different periods)
3. Run the analysis using:
   ```bash
   oq engine --run hazard.ini
   ```
4. Extract results (hazard curves, disaggregation, or UHS).

### **3. Using GEM’s Hazard API (For Developers)**
GEM provides APIs for programmatic access:
- **Global Hazard Mosaic API**: Retrieve hazard data via HTTP requests.
- Example (Python):
  ```python
  import requests
  response = requests.get("https://api.globalquakemodel.org/hazard/global/point?lat=34.05&lon=-118.25")
  print(response.json())  # Returns hazard data for Los Angeles
  ```

### **4. Download Pre-Computed Hazard Data**
- **GEM Hazard Data Portal**: Download pre-calculated hazard maps in GIS formats (Shapefile, GeoTIFF).
- **USGS & EMME (Europe)**: Some regions have detailed models integrated with GEM.

### **5. Validate & Compare**
- Cross-check with other models (e.g., **USGS Seismic Hazard Maps**, **EUCEM for Europe**).
- Use **OpenQuake’s disaggregation tool** to understand dominant earthquake scenarios.

### **Example Output**
For a location like **Tokyo (35.68°N, 139.77°E)**:
- **PGA for 475-year return period**: ~0.25g
- **Spectral Acceleration (1.0s)**: ~0.4g (for 2475-year return period)

# USGS $V_{s,30}$ Mapping Tool

The **USGS Vs30 Mapping Tool** is an online resource provided by the **United States Geological Survey (USGS)** that allows users to estimate the **average shear-wave velocity in the top 30 meters of the Earth's subsurface (Vs30)** at a specific location. Vs30 is a critical parameter in seismic hazard assessment, as it influences ground motion amplification during earthquakes.

### **Key Features of the USGS Vs30 Mapping Tool**
1. **Global Vs30 Data**  
   - Provides Vs30 estimates worldwide using different methodologies:
     - **Measured data** (from boreholes, geophysical surveys)
     - **Inferred data** (based on geology, topography, and remote sensing)
   - Includes datasets like:
     - **USGS Global Vs30 Map** (based on topographic slope)
     - **Detailed regional maps** (e.g., for the U.S., Japan, and other seismically active regions)

2. **Interactive Web Interface**  
   - Accessible at: [USGS Vs30 Map Viewer](https://earthquake.usgs.gov/data/vs30/)  
   - Users can:
     - Zoom into a location (latitude/longitude)
     - Click to get Vs30 values
     - Compare different Vs30 models

3. **Data Export Options**  
   - Download Vs30 estimates in formats like:
     - **CSV** (for coordinates and values)
     - **GeoTIFF** (for GIS applications)
     - **Web Map Services (WMS)** for integration into other tools

4. **Applications in Seismic Hazard Analysis**  
   - Used in:
     - **Site response studies** (to predict ground shaking)
     - **Building code development** (e.g., NEHRP site classification)
     - **Earthquake risk modeling** (e.g., ShakeMap, PAGER)

### **How to Use the USGS Vs30 Tool**
1. **Go to the USGS Vs30 Website**:  
   - Visit [https://earthquake.usgs.gov/data/vs30/](https://earthquake.usgs.gov/data/vs30/)

2. **Select a Vs30 Model**:  
   - Choose between:
     - **Global Vs30 based on topography** (default)
     - **Higher-resolution regional datasets** (if available)

3. **Find Your Location**:  
   - Enter coordinates or click on the map.
   - The tool will display the Vs30 value (in m/s) and the corresponding **NEHRP site class** (e.g., Site Class B, C, D).

4. **Download Data (Optional)**:  
   - Use the "Download" option to extract Vs30 data for further analysis.

### **Example Output**
For **Los Angeles, California (34.05°N, 118.25°W)**:  
- **Vs30 ≈ 350 m/s** (Site Class D - Stiff Soil)  
- **NEHRP Classification**: Used in building codes to determine seismic design forces.

### **Alternative Vs30 Estimation Methods**
If the USGS tool lacks resolution for your area, consider:
- **Local geophysical surveys** (e.g., MASW, ReMi seismic tests)
- **European Vs30 databases** (e.g., [ESDAC](https://esdac.jrc.ec.europa.eu/))
- **OpenQuake’s Vs30 datasets** (integrated with GEM models)


# MacOS Instructions

Yes! **OpenQuake** and related **GEM tools** can be used on **macOS**, though setup differs slightly from Windows/Linux. Here’s how to run them natively or via alternatives:

---

### **1. OpenQuake Engine on macOS**
#### **Option A: Native Installation (Recommended for Advanced Users)**
1. **Prerequisites**:
   - Install **Python 3.9+** (via [Homebrew](https://brew.sh/)):
     ```bash
     brew install python
     ```
   - Install dependencies:
     ```bash
     brew install gcc geos proj
     ```

2. **Install OpenQuake**:
   ```bash
   pip install openquake.engine
   ```

3. **Verify Installation**:
   ```bash
   oq engine --version
   ```

#### **Option B: Docker (Simpler, Cross-Platform)**
   ```bash
   docker pull gem/oq-engine
   docker run -it --rm -v $(pwd):/tmp gem/oq-engine bash
   ```
   - Run OpenQuake commands inside the container.

---

### **2. GEM Hazard Tools (Web-Based)**
- Tools like the **GEM Hazard Mosaic** or **OpenQuake Platform** are browser-based and work on any OS:
  - Access: [https://platform.openquake.org/](https://platform.openquake.org/)
  - No installation needed; export data directly (CSV, GeoJSON, PNG).

---

### **3. GIS Integration (QGIS on macOS)**
- **QGIS** (macOS native) can visualize OpenQuake outputs:
  1. Install [QGIS for macOS](https://qgis.org/en/site/forusers/download.html).
  2. Import OpenQuake-generated **Shapefiles/GeoTIFFs**.
  3. Style maps for reports.

---

### **4. Common macOS Issues & Fixes**
- **"Command not found: oq"**:
  - Fix: Ensure Python scripts are in your `PATH` (add `~/Library/Python/3.x/bin` to `.zshrc` or `.bash_profile`).
- **Library Errors (e.g., GEOS, PROJ)**:
  - Fix: Reinstall via Homebrew:
    ```bash
    brew reinstall geos proj
    ```

---

### **5. Alternatives for Lightweight Use**
- **USGS Tools** (e.g., [Vs30 Map Viewer](https://earthquake.usgs.gov/data/vs30/)) work in any browser.
- **Python Scripts** (using `openquake.hazardlib`) can be run locally if you avoid GUI dependencies.

---

### **6. Example: Running a Hazard Analysis on macOS**
```bash
# 1. Create a hazard.ini file
echo '[general]
description = macOS test
calculation_mode = classical
reference_vs30_type = measured
reference_vs30_value = 760
reference_depth_to_2pt5km_per_sec = 5.0

[site_params]
vs30 = 400
z1pt0 = 30

[erf]
rupture_mesh_spacing = 5
width_of_mfd_bin = 0.1
area_source_discretization = 10

[logic_tree]
number_of_logic_tree_samples = 0

[correlation_model]
type = JB2009

[intensity_measure_types_and_levels]
PGA = [0.01, 0.1, 0.2, 0.5, 1.0]

[site]
site_model_file = site_model.csv

[geometry]
region = -118.0, 34.0, -117.0, 35.0
region_grid_spacing = 0.01' > hazard.ini

# 2. Run the analysis
oq engine --run hazard.ini

# 3. Plot results (requires matplotlib)
oq plot hazard_curve.xml
```
